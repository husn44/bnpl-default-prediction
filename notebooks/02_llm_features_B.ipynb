# Load Data and Assign LLM Chat Messages & Product Descriptions

```python
import pandas as pd
import numpy as np

# Load datasets
df = pd.read_csv("raw_bnpl.csv")
chat_df = pd.read_csv("chat_messages.csv")
prod_df = pd.read_csv("product_texts.csv")

# Set fixed seed for reproducibility
rng = np.random.default_rng(42)

# Map support_chat and financial_stress_score matching _stress_hint
chat_groups = {lvl: grp.reset_index(drop=True) for lvl, grp in chat_df.groupby("stress_level")}
assigned_chats = []
assigned_stress_scores = []

for hint in df["_stress_hint"]:
    pool = chat_groups[hint]
    idx = rng.integers(0, len(pool))
    assigned_chats.append(pool.loc[idx, "message"])
    assigned_stress_scores.append(pool.loc[idx, "stress_score"])

df["support_chat"] = assigned_chats
df["financial_stress_score"] = assigned_stress_scores

# Map product_description and impulse_score matching _impulse_hint
prod_groups = {lvl: grp.reset_index(drop=True) for lvl, grp in prod_df.groupby("impulse_level")}
assigned_prods = []
assigned_impulse_scores = []

for hint in df["_impulse_hint"]:
    pool = prod_groups[hint]
    idx = rng.integers(0, len(pool))
    assigned_prods.append(pool.loc[idx, "description"])
    assigned_impulse_scores.append(pool.loc[idx, "impulse_score"])

df["product_description"] = assigned_prods
df["impulse_score"] = assigned_impulse_scores

# Sanity check
print(df[["_stress_hint", "support_chat", "financial_stress_score", "_impulse_hint", "product_description", "impulse_score"]].head(10))
print("\nMissing values in new columns:\n", df[["support_chat", "financial_stress_score", "product_description", "impulse_score"]].isnull().sum())
```
